## 3. Quick look at the data (2-3 real numbers)

To validate this lane for the next 7 weeks, I pulled some baseline numbers from the `data/raw/content_refresh_anonymized.csv` starter dataset, keeping in mind gotchas like `avg_position = 0` meaning "no data".

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

# Load the starter dataset
try:
    df = pd.read_csv('data/raw/content_refresh_anonymized.csv')

    total_rows = len(df)
    missing_rank = len(df[df['avg_position'] == 0]) # 0 means "no data", not rank 0
    avg_ctr = df['ctr'].mean()

    print(f"Total content items available: {total_rows:,}")
    print(f"Items with missing rank data (avg_position=0): {missing_rank:,}")
    print(f"Baseline Average CTR across the dataset: {avg_ctr:.2f}%")
except FileNotFoundError:
    print("Dataset not found. Please check the file path.")
```
